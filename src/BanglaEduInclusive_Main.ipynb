```python
from google.colab import userdata
OPENAI_API_KEY = userdata.get('OPEN_AI_API')
```

```python
import openai
import pandas as pd
import json
from langchain.chat_models import ChatOpenAI
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate
```

#Direct Prompting (Gender)

```python
# Define the prompt template
QA_DATASET_CATEGORIZATION_PROMPT = PromptTemplate.from_template(
    "{paragraph}\n"
    "Read the given paragraph and categorize it based on its inclusiveness of gender into one of the following categories:\n"
    "a) Inclusive\n"
    "b) Marginal Inclusive"
    "c) Marginal Non-inclusive"
    "d) Non-inclusive\n"
    "e) Neutral"
)
```

```python
# Load data from CSV
data = pd.read_csv("/content/paper_dataset.csv", header=None)
paragraphs = data[0].tolist()
```

```python
# Initialize the OpenAI language model
llm = ChatOpenAI(
    model="gpt-4o",
    api_key=OPENAI_API_KEY,
    temperature=0.7
)

# Create a chain with the LLM and the prompt template
chain = LLMChain(
    prompt=QA_DATASET_CATEGORIZATION_PROMPT,
    llm=llm
)
```

```python
# Process each paragraph through the chain
results = []
for paragraph in paragraphs:
    result = chain.run({"paragraph": paragraph})
    print(f"Paragraph: {paragraph}")
    print(f"Categorization: {result}\n")
    results.append({"paragraph": paragraph, "categorization": result})

# Convert the results into a DataFrame for better readability
results_df = pd.DataFrame(results)
print(results_df)

# Save the results to a CSV file
results_df.to_csv("categorized_inclusiveness_results.csv", index=False)
```
